```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Ecommerce_Unclean_Project.xlsx")
```

```python
#remove duplicate

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df
```

```python
df.replace(['N/A','NULL',''],np.nan)
```

```python
df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
df
```

```python
df = df[df['Email'].str.contains('@',na=False)]
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
df = df[df['Qty']>0]
```

```python
df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df
```

```python
df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df
```

```python
df['Net_Amount'] = df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
df['Profit'] = df['Net_Amount']*0.20
```

```python
df
```

```python
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
df['Year'] = df['Order_Date'].dt.year
```

```python
df
```

```python
df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.to_csv('Clean_Ecommerce.csv',index=False)
```

```python
from sqlalchemy import create_engine

server = r"localhost\SQLEXPRESS"
database = "Ecommerce"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?trusted_connection=yes&driver=ODBC+Driver+17+for+SQL+Server"
)

engine = create_engine(connection_string)

print("Connection created")
```

```python
with engine.connect() as conn:
    print("SQL Server connected successfully!")
```

```python
df.to_sql(
    "Clean_Ecommerce",
    con=engine,
    if_exists="replace",
    index=False
)

print("Data loaded into SQL Server!")
```
